# Analitic results the execution methods detection error 

This notebook report quality error detection for methods in SOTA

```python
import pandas as pd
```

```python
data = pd.read_csv("../outputs/results.csv")
print(data.shape)
data.head()
```

```python
results_raha = data[data['method'].str.contains('raha')]
print(results_raha.shape)
results_raha.head()
```

```python
results_raha.groupby('dataset').agg({'precision': ['mean', 'std'], 'recall': ['mean', 'std'], 'f1-score': ['mean', 'std']})
```

```python
results_dboost = data[data['method'].str.contains('dboost')]
print(results_dboost.shape)
results_dboost.head()
```

```python
results_dboost.groupby('dataset').agg({'precision': ['mean', 'std'], 'recall': ['mean', 'std'], 'f1-score': ['mean', 'std']})
```

```python
# Melhor F1-score e configuração correspondente para cada base de dados.
valid_results = results_raha.dropna(subset=['f1-score'])
best_indices = valid_results.groupby('dataset')['f1-score'].idxmax()
best_by_dataset = (
    valid_results.loc[best_indices, ['dataset', 'method','precision', 'recall', 'f1-score', 'config']]
    .sort_values('dataset')
    .reset_index(drop=True)
)
best_by_dataset
```

```python
# Melhor F1-score e configuração correspondente para cada base de dados.
valid_results = results_dboost.dropna(subset=['f1-score'])
best_indices = valid_results.groupby('dataset')['f1-score'].idxmax()
best_by_dataset = (
    valid_results.loc[best_indices, ['dataset', 'method','precision', 'recall', 'f1-score', 'config']]
    .sort_values('dataset')
    .reset_index(drop=True)
)
best_by_dataset
```

```python
# Melhor F1-score e configuração correspondente para cada base de dados.
valid_results = results_raha.dropna(subset=['f1-score'])
best_indices = valid_results.groupby('dataset')['f1-score'].idxmax()
best_by_dataset = (
    valid_results.loc[best_indices, ['dataset','f1-score', 'precision', 'recall',  'config']]
    .sort_values('dataset')
    .reset_index(drop=True)
)
for index, row in best_by_dataset.iterrows():
    print(row['dataset'], )
    print(row['config'])
```
