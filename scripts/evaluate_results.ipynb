```python
prices = {  # USD por 1M tokens
    "gpt-5-nano": {"input": 0.05, "output": 0.40},
}
```

```python
import json
llm = data[data["LLM"] == True].copy()
llm["llm_model"] = llm["config"].apply(lambda c: json.loads(c).get("llm_model"))

def cost(row):
    p = prices.get(row["llm_model"])
    if p is None:
        return float("nan")  # modelo sem preço (ex.: local)
    return row["input_tokens"] / 1e6 * p["input"] + row["output_tokens"] / 1e6 * p["output"]

llm["cost_usd"] = llm.apply(cost, axis=1)
llm[["dataset", "method", "input_tokens", "output_tokens", "cost_usd"]]
```
